````python
# ============================================================
# Cell 3 — Mount Drive, copy project files, download dataset
#
# Kaggle download SELALU dijalankan — kagglehub pakai lokal cache
# jika sudah pernah download, jadi cepat (tidak re-download).
# IMAGE_DIR selalu tersedia untuk Cell 4 dan Cell 12 (visualisasi).
#
# Yang di-skip jika cache ada:
#   - CSV split (jika TRAIN_CSV + TEST_CSV sudah ada di Drive)
#   - Landmark extraction (Cell 4 yang handle skip-nya)
# ============================================================
from google.colab import drive
drive.mount('/content/drive')

import os, sys, shutil, pandas as pd
from sklearn.model_selection import train_test_split

DRIVE_ROOT   = '/content/drive/MyDrive/Colab Notebooks/FaceRankNet/FaceRankNet'
PROJECT_ROOT = '/content/FaceRankNet'
CACHE_DIR    = f'{DRIVE_ROOT}/cache'
os.makedirs(PROJECT_ROOT, exist_ok=True)
os.makedirs(CACHE_DIR,    exist_ok=True)

# ── Copy .py files from Drive to runtime ────────────────────────────────────
copied = []
for fname in os.listdir(DRIVE_ROOT):
    if fname.endswith('.py'):
        shutil.copy(os.path.join(DRIVE_ROOT, fname), os.path.join(PROJECT_ROOT, fname))
        copied.append(fname)
sys.path.insert(0, PROJECT_ROOT)
print(f'Copied {len(copied)} .py files: {copied}')

# ── Define all paths ────────────────────────────────────────────────────────
TRAIN_CSV     = f'{DRIVE_ROOT}/train_labels.csv'
TEST_CSV      = f'{DRIVE_ROOT}/test_labels.csv'
CACHE_TRAIN   = f'{CACHE_DIR}/train_landmarks.pkl'
CACHE_TEST    = f'{CACHE_DIR}/test_landmarks.pkl'
AVG_FACE_PATH = f'{CACHE_DIR}/avg_face.npy'
PSEUDO_PATH   = f'{CACHE_DIR}/pseudo_labels.pkl'
CHECKPOINT    = f'{DRIVE_ROOT}/checkpoint_best.pt'

# ── Cache status check ───────────────────────────────────────────────────────
csvs_exist       = os.path.exists(TRAIN_CSV)   and os.path.exists(TEST_CSV)
landmarks_cached = os.path.exists(CACHE_TRAIN) and os.path.exists(CACHE_TEST)

print(f'\nCache status:')
print(f'  train_labels.csv    : {"\u2713" if os.path.exists(TRAIN_CSV) else "\u2717"}')
print(f'  test_labels.csv     : {"\u2713" if os.path.exists(TEST_CSV) else "\u2717"}')
print(f'  train_landmarks.pkl : {"\u2713" if os.path.exists(CACHE_TRAIN) else "\u2717"}')
print(f'  test_landmarks.pkl  : {"\u2713" if os.path.exists(CACHE_TEST) else "\u2717"}')
print(f'  avg_face.npy        : {"\u2713" if os.path.exists(AVG_FACE_PATH) else "\u2717"}')
print(f'  pseudo_labels.pkl   : {"\u2713" if os.path.exists(PSEUDO_PATH) else "\u2717"}')
print(f'  checkpoint_best.pt  : {"\u2713" if os.path.exists(CHECKPOINT) else "\u2717"}')

# ── Kaggle download (selalu dijalankan — lokal cache jika sudah ada) ─────────
import kagglehub
print('\nChecking Kaggle dataset (uses local cache if already downloaded)...')
KAGGLE_PATH = kagglehub.dataset_download(
    'pranavchandane/scut-fbp5500-v2-facial-beauty-scores')
print('Dataset path:', KAGGLE_PATH)

IMAGE_DIR   = os.path.join(KAGGLE_PATH, 'Images', 'Images')
LABELS_FILE = os.path.join(KAGGLE_PATH, 'labels.txt')

# Fallback: cari folder gambar dan labels.txt jika path standar tidak ada
if not os.path.isdir(IMAGE_DIR):
    for root, _, files in os.walk(KAGGLE_PATH):
        if any(f.lower().endswith(('.jpg', '.png')) for f in files):
            IMAGE_DIR = root; break
if not os.path.isfile(LABELS_FILE):
    for root, _, files in os.walk(KAGGLE_PATH):
        if 'labels.txt' in files:
            LABELS_FILE = os.path.join(root, 'labels.txt'); break

n_img = len([f for f in os.listdir(IMAGE_DIR) if f.lower().endswith(('.jpg', '.png'))])
print(f'\u2713 Image dir  : {IMAGE_DIR}  ({n_img} images)')
print(f'\u2713 Labels file: {LABELS_FILE}')

# ── Parse labels + train/test split (skip jika CSVs sudah ada) ───────────────
if not csvs_exist:
    df = pd.read_csv(LABELS_FILE, sep=' ', header=None, names=['Filename', 'Rating'])
    df['Ethnicity'] = df['Filename'].apply(
        lambda f: 'Asian' if f[0].upper() == 'A' else 'Caucasian')
    df['Gender'] = df['Filename'].apply(
        lambda f: 'Female' if f[1].upper() == 'F' else 'Male')
    print('Ethnicity dist:', df['Ethnicity'].value_counts().to_dict())
    print('Gender dist:',    df['Gender'].value_counts().to_dict())

    df['_b'] = df['Rating'].round(0).astype(int).clip(1, 5)
    train_df, test_df = train_test_split(
        df.drop(columns=['_b']), test_size=0.20, random_state=42, stratify=df['_b'])
    train_df = train_df.reset_index(drop=True)
    test_df  = test_df.reset_index(drop=True)
    train_df.to_csv(TRAIN_CSV, index=False)
    test_df.to_csv(TEST_CSV,   index=False)
    print(f'Train: {len(train_df)} | Test: {len(test_df)} \u2014 saved to Drive')
else:
    train_df = pd.read_csv(TRAIN_CSV)
    test_df  = pd.read_csv(TEST_CSV)
    # Backfill Ethnicity/Gender jika CSV di-save sebelum kolom ini ada
    _needs_resave = False
    for _df in [train_df, test_df]:
        if 'Ethnicity' not in _df.columns:
            _df['Ethnicity'] = _df['Filename'].apply(
                lambda f: 'Asian' if f[0].upper() == 'A' else 'Caucasian')
            _needs_resave = True
        if 'Gender' not in _df.columns:
            _df['Gender'] = _df['Filename'].apply(
                lambda f: 'Female' if f[1].upper() == 'F' else 'Male')
            _needs_resave = True
    if _needs_resave:
        train_df.to_csv(TRAIN_CSV, index=False)
        test_df.to_csv(TEST_CSV,   index=False)
        print('  \u2193 Backfilled Ethnicity/Gender columns & re-saved CSVs to Drive')
    print(f'\n\u2713 CSVs loaded dari Drive: Train={len(train_df)} | Test={len(test_df)}')

print(f'\nAll paths ready.')
print(f'  IMAGE_DIR  : {IMAGE_DIR}')
print(f'  CHECKPOINT : {CHECKPOINT}')
````

````python
# ============================================================
# Cell 5 — Pseudo-Labels via beauty-axis projection (per-ethnicity H1)
# Skips compute jika avg_face + pseudo_labels sudah ada di Drive.
# ============================================================
import os, pickle, numpy as np, pandas as pd
from pseudo_labels import (
    compute_universal_average_face,
    compute_beauty_prototype,
    compute_ethnicity_avg_faces,
    compute_all_pseudo_labels_beauty_axis,
    validate_pseudo_label_quality,
    save_avg_face, save_pseudo_labels,
    load_pseudo_labels, load_avg_face,
)

train_df_loaded = pd.read_csv(TRAIN_CSV)
train_filenames = train_df_loaded["Filename"].tolist()
holistic_map    = dict(zip(train_df_loaded["Filename"], train_df_loaded["Rating"].astype(float)))
ethnicity_map   = dict(zip(train_df_loaded["Filename"], train_df_loaded["Ethnicity"]))

cache_hit = os.path.exists(AVG_FACE_PATH) and os.path.exists(PSEUDO_PATH)

if cache_hit:
    print('✓ Caches found — loading from Drive (skip compute)')
    population_mean = load_avg_face(AVG_FACE_PATH)
    pseudo_labels   = load_pseudo_labels(PSEUDO_PATH)
else:
    if 'train_coords' not in globals():
        with open(CACHE_TRAIN, 'rb') as f:
            train_coords = pickle.load(f)

    train_coords_list  = [train_coords[f] for f in train_filenames if f in train_coords]
    train_ratings_list = [holistic_map[f]  for f in train_filenames if f in train_coords]

    population_mean  = compute_universal_average_face(train_coords_list)
    beauty_prototype = compute_beauty_prototype(train_coords_list, train_ratings_list, top_k_pct=0.30)

    pop_mean_eth_map     = compute_ethnicity_avg_faces(
        train_coords, train_filenames, ethnicity_map, holistic_ratings=None)
    beauty_proto_eth_map = compute_ethnicity_avg_faces(
        train_coords, train_filenames, ethnicity_map,
        holistic_ratings=holistic_map, top_k_pct=0.30)

    print('Computing beauty-axis pseudo-labels (per-ethnicity) ...')
    pseudo_labels = compute_all_pseudo_labels_beauty_axis(
        coords_cache=train_coords,
        train_filenames=train_filenames,
        population_mean=population_mean,
        beauty_prototype=beauty_prototype,
        population_mean_map=pop_mean_eth_map,
        beauty_prototype_map=beauty_proto_eth_map,
        ethnicity_map=ethnicity_map,
    )

    save_avg_face(population_mean, AVG_FACE_PATH)
    save_pseudo_labels(pseudo_labels, PSEUDO_PATH)
    print(f'✓ Saved: {AVG_FACE_PATH}')
    print(f'         {PSEUDO_PATH}')

rho = validate_pseudo_label_quality(pseudo_labels, holistic_map)
print(f'\nSpearman rho (beauty-axis): {rho:.4f}  (target > 0.2)')

print()
for sample in ["CF137.jpg", "CF581.jpg", "AF1859.jpg", "AF549.jpg"]:
    if sample in pseudo_labels:
        organs = pseudo_labels[sample]
        mean_score = sum(organs.values()) / len(organs)
        print(f'{sample}  rating={holistic_map.get(sample, "?"):.3f}  mean_pseudo={mean_score:.3f}')
        for organ, score in organs.items():
            print(f'  {organ:<12}: {score:.3f}')
````

````python
# ============================================================
# Cell 6 — Dataset & DataLoader
# ============================================================
import os, pickle, numpy as np
from pseudo_labels import load_pseudo_labels
from dataset import (
    FaceDataset, PairDataset,
    make_face_loader, make_weighted_pair_loader,
)
import config

if 'train_coords' not in globals():
    with open(CACHE_TRAIN, 'rb') as f: train_coords = pickle.load(f)
    print(f'✓ Loaded train_coords ({len(train_coords)} faces)')
if 'test_coords' not in globals():
    with open(CACHE_TEST, 'rb') as f: test_coords = pickle.load(f)
    print(f'✓ Loaded test_coords ({len(test_coords)} faces)')
if 'pseudo_labels' not in globals():
    pseudo_labels = load_pseudo_labels(PSEUDO_PATH)
    print(f'✓ Loaded pseudo_labels ({len(pseudo_labels)} entries)')
if 'avg_face' not in globals():
    avg_face = np.load(AVG_FACE_PATH)
    print(f'✓ Loaded avg_face shape={avg_face.shape}')

train_face_ds = FaceDataset(
    TRAIN_CSV, train_coords, pseudo_labels, avg_face=avg_face,
    augment_jitter=config.AUGMENT_JITTER,
)
test_face_ds = FaceDataset(TEST_CSV, test_coords, avg_face=avg_face)
pair_ds      = PairDataset(
    train_face_ds,
    hard_pair_sampling=config.USE_HARD_PAIR_SAMPLING,
)

pair_loader = make_weighted_pair_loader(pair_ds)
test_loader = make_face_loader(test_face_ds, shuffle=False)

print(f'\nTrain faces: {len(train_face_ds)} | Pairs: {len(pair_ds)} | Test: {len(test_face_ds)}')

ratings = np.array(train_face_ds.ratings)
pair_anchor_ratings = np.array([ratings[a] for a, _ in pair_ds._pairs])
counts_raw = np.histogram(pair_anchor_ratings, bins=[0, 2, 3, 4, 5.1])[0]
print(f'Pair anchor distribution (raw): {counts_raw.tolist()}  [<2, 2-3, 3-4, >4]')

batch_a, batch_b, mask = next(iter(pair_loader))
print('Batch A ratings shape :', batch_a['ratings'].shape)
print('Organ mask shape      :', mask.shape)
````

````python
# ============================================================
# Cell 10 — Visualisation: Organ Bar Chart + Cross-Organ Attention
#            (IEEE paper style)
# ============================================================
import torch, matplotlib.pyplot as plt, matplotlib, numpy as np
from dataset import collate_faces
from torch.utils.data import DataLoader

# ── IEEE paper style ─────────────────────────────────────────────────────────
matplotlib.rcParams.update({
    'font.family'      : 'serif',
    'font.serif'       : ['Times New Roman', 'DejaVu Serif', 'Palatino'],
    'font.size'        : 9,
    'axes.labelsize'   : 9,
    'axes.titlesize'   : 9,
    'xtick.labelsize'  : 8,
    'ytick.labelsize'  : 8,
    'legend.fontsize'  : 8,
    'legend.frameon'   : True,
    'legend.framealpha': 0.9,
    'legend.edgecolor' : '0.8',
    'axes.linewidth'   : 0.7,
    'axes.spines.top'  : False,
    'axes.spines.right': False,
    'xtick.major.width': 0.7,
    'ytick.major.width': 0.7,
    'lines.linewidth'  : 1.0,
    'patch.linewidth'  : 0.5,
    'grid.linewidth'   : 0.4,
    'grid.color'       : '0.88',
    'figure.dpi'       : 150,
    'savefig.dpi'      : 300,
    'savefig.bbox'     : 'tight',
})

ORGAN_COLORS = ['#2166AC', '#4393C3', '#4DAC26', '#D6604D', '#762A83']

sample_batch = next(iter(DataLoader(
    test_face_ds, batch_size=4, shuffle=False, collate_fn=collate_faces)))

trained_model.eval()
with torch.no_grad():
    sg  = {k: v.to(device) for k, v in sample_batch['subgraphs'].items()}
    out = trained_model(sg)

organ_names = list(out['local_scores'].keys())
n = len(sample_batch['filenames'])

# ── Fig 1: Organ score bar chart (one subplot per face) ──────────────────────
fig, axes = plt.subplots(1, n, figsize=(7.16, 2.8), sharey=True)
if n == 1: axes = [axes]

for i, ax in enumerate(axes):
    scores = [out['local_scores'][o][i].item() for o in organ_names]
    bars = ax.bar(range(len(organ_names)), scores,
                  color=ORGAN_COLORS[:len(organ_names)], linewidth=0, width=0.6, zorder=3)
    for bar, sc in zip(bars, scores):
        ax.text(bar.get_x() + bar.get_width() / 2, sc + 0.08,
                f'{sc:.2f}', ha='center', va='bottom', fontsize=7)
    g_score = out['global_score'][i].item()
    gt_val  = sample_batch['ratings'][i].item()
    ax.axhline(g_score, color='black', ls='--', lw=0.9, zorder=4,
               label=f'Global = {g_score:.2f}')
    ax.axhline(gt_val,  color='0.45',  ls=':',  lw=0.9, zorder=4,
               label=f'GT = {gt_val:.2f}')
    ax.set_ylim(1, 5.4)
    ax.set_xticks(range(len(organ_names)))
    short_labels = [o.replace('_eye', '\nEye').replace('_', '\n').title()
                    for o in organ_names]
    ax.set_xticklabels(short_labels, fontsize=7)
    ax.set_title(f'({chr(ord("a") + i)}) {sample_batch["filenames"][i]}',
                 loc='left', fontsize=8)
    ax.legend(fontsize=6.5, handlelength=1.5)
    ax.yaxis.grid(True, zorder=0); ax.set_axisbelow(True)

axes[0].set_ylabel('Beauty Score')
fig.suptitle('FaceRankNet — Organ-Level Beauty Scores (Sample Faces)', fontsize=9, y=1.02)
plt.tight_layout(pad=0.8)
plt.savefig(f'{DRIVE_ROOT}/organ_scores_bar.pdf')
plt.savefig(f'{DRIVE_ROOT}/organ_scores_bar.png')
plt.show(); plt.close()

# ── Fig 2 / 3: Cross-organ attention ─────────────────────────────────────────
if out['attn_weights'] is None:
    print("fusion_weight mode — no cross-organ attention heatmap.")
else:
    attn      = out['attn_weights'].cpu().numpy()
    attn_mean = attn.mean(axis=0)
    short_names = ['L.Eye', 'R.Eye', 'Nose', 'Mouth', 'Jaw']

    # Fig 2: Heatmap
    fig2, ax2 = plt.subplots(figsize=(3.2, 2.8))
    im = ax2.imshow(attn_mean, cmap='Blues', vmin=0)
    ax2.set_xticks(range(len(organ_names))); ax2.set_xticklabels(short_names, fontsize=7)
    ax2.set_yticks(range(len(organ_names))); ax2.set_yticklabels(short_names, fontsize=7)
    ax2.set_xlabel('Key (attended organ)', fontsize=8)
    ax2.set_ylabel('Query (attending organ)', fontsize=8)
    for i in range(len(organ_names)):
        for j in range(len(organ_names)):
            txt_c = 'white' if attn_mean[i, j] > attn_mean.max() * 0.6 else 'black'
            ax2.text(j, i, f'{attn_mean[i, j]:.3f}', ha='center', va='center',
                     fontsize=6.5, color=txt_c)
    cbar = plt.colorbar(im, ax=ax2, shrink=0.85)
    cbar.ax.tick_params(labelsize=7)
    ax2.set_title(f'Cross-Organ Attention ($N$={n} faces)', fontsize=9)
    plt.tight_layout()
    plt.savefig(f'{DRIVE_ROOT}/cross_organ_attention.pdf')
    plt.savefig(f'{DRIVE_ROOT}/cross_organ_attention.png')
    plt.show(); plt.close()

    # Fig 3: Attention importance bar chart
    received = attn_mean.sum(axis=0)
    eff_w    = received / received.sum()
    fig3, ax3 = plt.subplots(figsize=(3.5, 2.6))
    bars3 = ax3.bar(range(len(organ_names)), eff_w,
                    color=ORGAN_COLORS[:len(organ_names)], linewidth=0, width=0.55, zorder=3)
    for bar, v in zip(bars3, eff_w):
        ax3.text(bar.get_x() + bar.get_width() / 2, v + 0.003,
                 f'{v:.1%}', ha='center', va='bottom', fontsize=8)
    ax3.axhline(1 / len(organ_names), color='0.5', lw=0.8, ls='--',
                label=f'Uniform ({1/len(organ_names):.3f})')
    ax3.set_xticks(range(len(organ_names)))
    ax3.set_xticklabels(short_names, fontsize=8)
    ax3.set_ylabel('Normalized Attention Share')
    ax3.set_ylim(0, max(eff_w) * 1.3)
    ax3.set_title('Effective Organ Importance via Cross-Organ Attention\n'
                  r'Column-sum of $\bar{A}$ (averaged over batch)',
                  fontsize=9, loc='left')
    ax3.legend(fontsize=7, handlelength=1.5)
    ax3.yaxis.grid(True, zorder=0); ax3.set_axisbelow(True)
    plt.tight_layout()
    plt.savefig(f'{DRIVE_ROOT}/organ_importance_attention.pdf')
    plt.savefig(f'{DRIVE_ROOT}/organ_importance_attention.png')
    plt.show(); plt.close()

print('Figures saved to Drive (PDF + PNG).')
````

````python
# ============================================================
# Cell 11 — Extended Evaluation (IEEE paper style)
# 1. Scatter predicted vs GT
# 2. Error distribution per rating bucket
# 3. PCC/MAE per ethnicity (H1 check)
# 4. Score distribution KDE
# 5. Top-K ranking precision
# 6. Full test-set attention heatmap (skipped in fusion_weight mode)
# 7. Organ fusion weights (always shown)
# ============================================================
import torch, numpy as np, pandas as pd, matplotlib.pyplot as plt
import matplotlib
from scipy import stats
from torch.utils.data import DataLoader
from dataset import collate_faces

# ── IEEE paper style ──────────────────────────────────────────────────────────
matplotlib.rcParams.update({
    'font.family'      : 'serif',
    'font.serif'       : ['Times New Roman', 'DejaVu Serif', 'Palatino'],
    'font.size'        : 9,
    'axes.labelsize'   : 9,
    'axes.titlesize'   : 9,
    'xtick.labelsize'  : 8,
    'ytick.labelsize'  : 8,
    'legend.fontsize'  : 8,
    'legend.frameon'   : True,
    'legend.framealpha': 0.9,
    'legend.edgecolor' : '0.8',
    'axes.linewidth'   : 0.7,
    'axes.spines.top'  : False,
    'axes.spines.right': False,
    'xtick.major.width': 0.7,
    'ytick.major.width': 0.7,
    'lines.linewidth'  : 1.0,
    'patch.linewidth'  : 0.5,
    'grid.linewidth'   : 0.4,
    'grid.color'       : '0.88',
    'figure.dpi'       : 150,
    'savefig.dpi'      : 300,
    'savefig.bbox'     : 'tight',
})

C_BLUE   = '#2166AC'
C_RED    = '#D6604D'
C_GREEN  = '#4DAC26'
C_PURPLE = '#762A83'
C_GRAY   = '#555555'
C_ORANGE = '#F46D43'

# ── Collect all predictions on test set ─────────────────────────────────────
trained_model.eval()
all_preds, all_gts, all_files = [], [], []
all_attn = []

with torch.no_grad():
    for batch in DataLoader(test_face_ds, batch_size=64, shuffle=False, collate_fn=collate_faces):
        sg  = {k: v.to(device) for k, v in batch['subgraphs'].items()}
        out = trained_model(sg)
        all_preds.extend(out['global_score'].cpu().numpy())
        all_gts.extend(batch['ratings'].numpy())
        all_files.extend(batch['filenames'])
        if out['attn_weights'] is not None:
            all_attn.append(out['attn_weights'].cpu().numpy())

preds  = np.array(all_preds)
gts    = np.array(all_gts)
files  = np.array(all_files)
errors = preds - gts

# ── 1. Scatter plot: Predicted vs GT ────────────────────────────────────────
fig, ax = plt.subplots(figsize=(3.5, 3.5))
ax.scatter(gts, preds, s=6, alpha=0.4, color=C_BLUE, linewidths=0, rasterized=True)
ax.plot([1, 5], [1, 5], color=C_RED, lw=0.9, ls='--', label='Ideal')
m, b, r, _, _ = stats.linregress(gts, preds)
xs = np.linspace(1, 5, 200)
ax.plot(xs, m*xs + b, color=C_ORANGE, lw=1.0, label=f'Fit ($r$={r:.3f})')
pcc = float(np.corrcoef(gts, preds)[0, 1])
ax.set_xlabel('Ground-Truth Rating')
ax.set_ylabel('Predicted Score')
ax.set_title(f'Predicted vs. Ground Truth (PCC = {pcc:.4f})')
ax.set_xlim(1, 5); ax.set_ylim(1, 5)
ax.legend(loc='upper left')
ax.set_aspect('equal')
plt.tight_layout()
plt.savefig(f'{DRIVE_ROOT}/scatter_pred_vs_gt.pdf')
plt.savefig(f'{DRIVE_ROOT}/scatter_pred_vs_gt.png')
plt.show(); plt.close()

# ── 2. Error distribution per rating bucket ──────────────────────────────────
bucket_defs = [
    ('Unattractive\n(1–2)', 1, 2, C_BLUE),
    ('Average\n(2–3)',      2, 3, C_GREEN),
    ('Attractive\n(3–4)',   3, 4, C_ORANGE),
    ('Beautiful\n(4–5)',    4, 5, C_RED),
]
fig, axes = plt.subplots(1, 4, figsize=(7.16, 2.6), sharey=True)
for ax, (label, lo, hi, c) in zip(axes, bucket_defs):
    mask = (gts >= lo) & (gts < hi)
    err  = errors[mask]
    ax.hist(err, bins=18, color=c, alpha=0.85, linewidth=0)
    ax.axvline(0,          color='black', lw=0.8, ls='--')
    ax.axvline(err.mean(), color=C_RED,   lw=0.8, ls='-',
               label=f'$\\mu$={err.mean():.2f}')
    ax.set_title(f'{label}\n($n$={mask.sum()})', fontsize=8)
    ax.set_xlabel('Error (pred $-$ GT)', fontsize=8)
    ax.legend(handlelength=1.2)
axes[0].set_ylabel('Count')
fig.suptitle('Prediction Error Distribution by Rating Bucket', fontsize=9, y=1.02)
plt.tight_layout()
plt.savefig(f'{DRIVE_ROOT}/error_per_bucket.pdf')
plt.savefig(f'{DRIVE_ROOT}/error_per_bucket.png')
plt.show(); plt.close()

# ── 3. PCC & MAE per ethnicity (H1 evaluation) ───────────────────────────────
test_df_loaded = pd.read_csv(TEST_CSV)
eth_map = dict(zip(test_df_loaded['Filename'], test_df_loaded['Ethnicity']))
eth_data = {'Asian': ([], []), 'Caucasian': ([], [])}
for f, p, g in zip(files, preds, gts):
    eth = eth_map.get(f, '')
    if eth in eth_data:
        eth_data[eth][0].append(p)
        eth_data[eth][1].append(g)

eth_names, eth_pcc, eth_mae, eth_n = [], [], [], []
for eth, (ps, gs_e) in eth_data.items():
    if not ps: continue
    ps, gs_e = np.array(ps), np.array(gs_e)
    eth_names.append(eth)
    eth_pcc.append(float(np.corrcoef(ps, gs_e)[0, 1]))
    eth_mae.append(float(np.abs(ps - gs_e).mean()))
    eth_n.append(len(ps))

x  = np.arange(len(eth_names))
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(3.5, 2.8))
colors_eth = [C_BLUE, C_RED]

bars1 = ax1.bar(x, eth_pcc, width=0.7, color=colors_eth[:len(x)], linewidth=0)
ax1.axhline(0.70, color=C_GRAY, lw=0.8, ls='--', label='Target (0.70)')
ax1.set_xticks(x); ax1.set_xticklabels(eth_names)
ax1.set_ylim(0, 1.05); ax1.set_ylabel('PCC')
ax1.set_title('PCC by Ethnicity')
for bar, v, n in zip(bars1, eth_pcc, eth_n):
    ax1.text(bar.get_x() + bar.get_width()/2, v + 0.02,
             f'{v:.3f}\n($n$={n})', ha='center', va='bottom', fontsize=7)
ax1.legend(fontsize=7)

bars2 = ax2.bar(x, eth_mae, width=0.7, color=colors_eth[:len(x)], linewidth=0)
ax2.axhline(0.36, color=C_GRAY, lw=0.8, ls='--', label='Target (0.36)')
ax2.set_xticks(x); ax2.set_xticklabels(eth_names)
ax2.set_ylabel('MAE'); ax2.set_title('MAE by Ethnicity')
for bar, v in zip(bars2, eth_mae):
    ax2.text(bar.get_x() + bar.get_width()/2, v + 0.005,
             f'{v:.3f}', ha='center', va='bottom', fontsize=7)
ax2.legend(fontsize=7)

fig.suptitle('Per-Ethnicity Performance (H1 Validation)', fontsize=9)
plt.tight_layout()
plt.savefig(f'{DRIVE_ROOT}/ethnicity_metrics.pdf')
plt.savefig(f'{DRIVE_ROOT}/ethnicity_metrics.png')
plt.show(); plt.close()

# ── 4. Score distribution KDE ────────────────────────────────────────────────
from scipy.stats import gaussian_kde
fig, ax = plt.subplots(figsize=(3.5, 2.6))
xs = np.linspace(1, 5, 300)
for arr, label, c, ls in [
    (gts,   'Ground Truth', C_BLUE,   '-'),
    (preds, 'Predicted',    C_RED,    '--'),
]:
    kde = gaussian_kde(arr, bw_method=0.3)
    ax.plot(xs, kde(xs), color=c, lw=1.2, ls=ls, label=label)
    ax.fill_between(xs, kde(xs), alpha=0.10, color=c)
ax.set_xlabel('Score'); ax.set_ylabel('Density')
ax.set_title('Score Distribution')
ax.set_xlim(1, 5); ax.legend()
plt.tight_layout()
plt.savefig(f'{DRIVE_ROOT}/score_distribution_kde.pdf')
plt.savefig(f'{DRIVE_ROOT}/score_distribution_kde.png')
plt.show(); plt.close()

# ── 5. Top-K ranking precision ───────────────────────────────────────────────
print("Top-K Ranking Precision")
print(f"{'K':>5}  {'Precision':>10}")
print("-" * 18)
gt_rank   = np.argsort(-gts)
pred_rank = np.argsort(-preds)
for k in [10, 25, 50, 100]:
    prec = len(set(gt_rank[:k]) & set(pred_rank[:k])) / k
    print(f"{k:>5}  {prec:>10.1%}")

# ── 6. Cross-organ attention heatmap (skip in fusion_weight mode) ────────────
organ_names_short = ['Left Eye', 'Right Eye', 'Nose', 'Mouth', 'Jawline']
short_names       = ['L.Eye', 'R.Eye', 'Nose', 'Mouth', 'Jaw']

if all_attn:
    attn_all  = np.concatenate(all_attn, axis=0)
    attn_mean = attn_all.mean(axis=0)

    fig, ax = plt.subplots(figsize=(3.2, 2.8))
    im = ax.imshow(attn_mean, cmap='Blues', vmin=0)
    ax.set_xticks(range(5)); ax.set_xticklabels(short_names, fontsize=7)
    ax.set_yticks(range(5)); ax.set_yticklabels(short_names, fontsize=7)
    ax.set_xlabel('Key (attended organ)', fontsize=8)
    ax.set_ylabel('Query (attending organ)', fontsize=8)
    for i in range(5):
        for j in range(5):
            txt_c = 'white' if attn_mean[i, j] > attn_mean.max() * 0.6 else 'black'
            ax.text(j, i, f'{attn_mean[i,j]:.3f}', ha='center', va='center',
                    fontsize=6.5, color=txt_c)
    cbar = plt.colorbar(im, ax=ax, shrink=0.85)
    cbar.ax.tick_params(labelsize=7)
    ax.set_title(f'Cross-Organ Attention ($N$={len(attn_all)} faces)', fontsize=9)
    plt.tight_layout()
    plt.savefig(f'{DRIVE_ROOT}/attention_fulltest.pdf')
    plt.savefig(f'{DRIVE_ROOT}/attention_fulltest.png')
    plt.show(); plt.close()
else:
    print('\n(fusion_weight mode — cross-organ attention heatmap not applicable)')

# ── 7. Organ fusion weights (always shown in fusion_weight mode) ─────────────
if hasattr(trained_model, 'fusion_weights'):
    fw_raw     = trained_model.fusion_weights.detach().cpu()
    fw_softmax = torch.softmax(fw_raw, dim=0).numpy()

    organ_colors = ['#2166AC', '#4393C3', '#4DAC26', '#D6604D', '#762A83']
    x = np.arange(len(organ_names_short))

    if all_attn:
        # Both subplots: learned fusion weights + attention-based effective weight
        attn_col_sum = attn_mean.sum(axis=0)
        attn_eff_w   = attn_col_sum / attn_col_sum.sum()
        fig, (ax_fw, ax_attn_w) = plt.subplots(1, 2, figsize=(7.16, 2.6))

        bars_fw = ax_fw.bar(x, fw_softmax, color=organ_colors, linewidth=0, width=0.55, zorder=3)
        for bar, v in zip(bars_fw, fw_softmax):
            ax_fw.text(bar.get_x() + bar.get_width() / 2, v + 0.003,
                       f'{v:.3f}', ha='center', va='bottom', fontsize=8)
        ax_fw.axhline(1/5, color='0.5', lw=0.8, ls='--', label='Uniform (0.200)')
        ax_fw.set_xticks(x); ax_fw.set_xticklabels(organ_names_short, rotation=15, ha='right', fontsize=8)
        ax_fw.set_ylabel('Weight'); ax_fw.set_ylim(0, max(fw_softmax) * 1.25)
        ax_fw.set_title('(a) Learned Organ Fusion Weights\n'
                        r'$\mathbf{w} = \mathrm{softmax}(\theta_{\mathrm{fuse}})$',
                        fontsize=9, loc='left')
        ax_fw.legend(fontsize=7, handlelength=1.5)
        ax_fw.yaxis.grid(True, zorder=0); ax_fw.set_axisbelow(True)

        bars_aw = ax_attn_w.bar(x, attn_eff_w, color=organ_colors, linewidth=0, width=0.55, zorder=3)
        for bar, v in zip(bars_aw, attn_eff_w):
            ax_attn_w.text(bar.get_x() + bar.get_width() / 2, v + 0.003,
                           f'{v:.3f}', ha='center', va='bottom', fontsize=8)
        ax_attn_w.axhline(1/5, color='0.5', lw=0.8, ls='--', label='Uniform (0.200)')
        ax_attn_w.set_xticks(x); ax_attn_w.set_xticklabels(organ_names_short, rotation=15, ha='right', fontsize=8)
        ax_attn_w.set_ylabel('Normalized Attention Share'); ax_attn_w.set_ylim(0, max(attn_eff_w) * 1.25)
        ax_attn_w.set_title('(b) Effective Organ Weight via Cross-Organ Attention\n'
                            r'Column-sum of $\bar{A}$ (averaged over test set)',
                            fontsize=9, loc='left')
        ax_attn_w.legend(fontsize=7, handlelength=1.5)
        ax_attn_w.yaxis.grid(True, zorder=0); ax_attn_w.set_axisbelow(True)
    else:
        # fusion_weight mode: only show learned fusion weights
        fig, ax_fw = plt.subplots(figsize=(4.0, 2.6))
        bars_fw = ax_fw.bar(x, fw_softmax, color=organ_colors, linewidth=0, width=0.55, zorder=3)
        for bar, v in zip(bars_fw, fw_softmax):
            ax_fw.text(bar.get_x() + bar.get_width() / 2, v + 0.003,
                       f'{v:.3f}', ha='center', va='bottom', fontsize=8)
        ax_fw.axhline(1/5, color='0.5', lw=0.8, ls='--', label='Uniform (0.200)')
        ax_fw.set_xticks(x); ax_fw.set_xticklabels(organ_names_short, rotation=15, ha='right', fontsize=8)
        ax_fw.set_ylabel('Weight'); ax_fw.set_ylim(0, max(fw_softmax) * 1.25)
        ax_fw.set_title('Learned Organ Fusion Weights\n'
                        r'$\mathbf{w} = \mathrm{softmax}(\theta_{\mathrm{fuse}})$',
                        fontsize=9)
        ax_fw.legend(fontsize=7, handlelength=1.5)
        ax_fw.yaxis.grid(True, zorder=0); ax_fw.set_axisbelow(True)

    plt.tight_layout(pad=0.8)
    plt.savefig(f'{DRIVE_ROOT}/organ_fusion_weights.pdf')
    plt.savefig(f'{DRIVE_ROOT}/organ_fusion_weights.png')
    plt.show(); plt.close()

    print('\nOrgan Fusion Weights Summary:')
    print(f'  {"Organ":<12}  {"Learned w":>10}')
    print('  ' + '-' * 26)
    for name, lw in zip(organ_names_short, fw_softmax):
        marker = ' *' if lw == fw_softmax.max() else ''
        print(f'  {name:<12}  {lw:>10.4f}{marker}')
    print('  (* highest learned weight)')

print('\nAll figures saved to Drive (PDF + PNG).')
````

````python
# ============================================================
# Cell 12 — Visualisasi Final: Wajah + Bounding Box Organ + Skor Bar Chart
#            (IEEE paper style — bounding box tanpa label teks)
#
# Prerequisite: Cell 8 atau 8b (model loaded), test_face_ds dari Cell 6.
# ============================================================
import os, cv2, numpy as np, torch, matplotlib.pyplot as plt
import matplotlib, matplotlib.patches as mpatches
import pandas as pd
import mediapipe as mp
from dataset import collate_faces
from organ_indices import ORGAN_INDICES

# ── IEEE paper style ──────────────────────────────────────────────────────────
matplotlib.rcParams.update({
    'font.family'      : 'serif',
    'font.serif'       : ['Times New Roman', 'DejaVu Serif', 'Palatino'],
    'font.size'        : 9,
    'axes.labelsize'   : 9,
    'axes.titlesize'   : 9,
    'xtick.labelsize'  : 8,
    'ytick.labelsize'  : 8,
    'legend.fontsize'  : 8,
    'legend.frameon'   : True,
    'legend.framealpha': 0.9,
    'legend.edgecolor' : '0.8',
    'axes.linewidth'   : 0.7,
    'axes.spines.top'  : False,
    'axes.spines.right': False,
    'xtick.major.width': 0.7,
    'ytick.major.width': 0.7,
    'grid.linewidth'   : 0.4,
    'grid.color'       : '0.88',
    'figure.dpi'       : 150,
    'savefig.dpi'      : 300,
    'savefig.bbox'     : 'tight',
})

# ── Konfigurasi ───────────────────────────────────────────────────────────────
N_SAMPLES   = 4
RANDOM_SEED = 7

# ── MediaPipe: support legacy API (<0.10.15) dan Tasks API (>=0.10.15) ────────
_USE_LEGACY_MP = hasattr(mp, 'solutions') and hasattr(mp.solutions, 'face_mesh')

if _USE_LEGACY_MP:
    _fmesh_viz = mp.solutions.face_mesh.FaceMesh(
        static_image_mode=True, max_num_faces=1,
        refine_landmarks=False, min_detection_confidence=0.5,
    )
    print('\u2713 MediaPipe legacy API')
else:
    import urllib.request
    from mediapipe.tasks import python as _mp_python
    from mediapipe.tasks.python import vision as _mp_vision
    _MODEL_URL  = ('https://storage.googleapis.com/mediapipe-models/'
                   'face_landmarker/face_landmarker/float16/1/face_landmarker.task')
    _model_path = '/tmp/face_landmarker.task'
    if not os.path.exists(_model_path):
        print('Downloading MediaPipe FaceLandmarker model (~30 MB)...')
        urllib.request.urlretrieve(_MODEL_URL, _model_path)
    _fmesh_viz = _mp_vision.FaceLandmarker.create_from_options(
        _mp_vision.FaceLandmarkerOptions(
            base_options=_mp_python.BaseOptions(model_asset_path=_model_path),
            num_faces=1,
            min_face_detection_confidence=0.5,
        )
    )
    print('\u2713 MediaPipe Tasks API')

# Warna: muted ColorBrewer qualitative palette
_COLORS_BGR = {
    'left_eye':  ( 33, 102, 172),   # biru tua   — BGR
    'right_eye': ( 67, 147, 195),   # biru muda
    'nose':      ( 26, 152,  80),   # hijau
    'mouth':     ( 49,  54, 149),   # merah gelap (BGR: b=D6, g=60, r=4D → 4D,60,D6)
    'jawline':   (118,  42, 131),   # ungu
}
_COLORS_BGR['mouth']   = ( 77,  96, 214)   # BGR of #D6604D
_COLORS_BGR['jawline'] = (131,  42, 118)   # BGR of #762A83

_COLORS_HEX = {
    'left_eye':  '#2166AC',
    'right_eye': '#4393C3',
    'nose':      '#4DAC26',
    'mouth':     '#D6604D',
    'jawline':   '#762A83',
}
_ORGAN_LABELS = {
    'left_eye':  'Left Eye',
    'right_eye': 'Right Eye',
    'nose':      'Nose',
    'mouth':     'Mouth',
    'jawline':   'Jawline',
}

def _get_pixel_landmarks(image_path: str):
    """Return (rgb_img, coords_px (468,2)) atau None jika wajah tak terdeteksi."""
    bgr = cv2.imread(image_path)
    if bgr is None:
        return None
    rgb = cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB)
    h, w = rgb.shape[:2]
    if _USE_LEGACY_MP:
        res = _fmesh_viz.process(rgb)
        if not res.multi_face_landmarks:
            return None
        lm = res.multi_face_landmarks[0].landmark
        coords_norm = np.array([[p.x, p.y] for p in lm], dtype=np.float32)[:468]
    else:
        mp_img = mp.Image(image_format=mp.ImageFormat.SRGB, data=rgb)
        res    = _fmesh_viz.detect(mp_img)
        if not res.face_landmarks:
            return None
        lm = res.face_landmarks[0]
        coords_norm = np.array([[p.x, p.y] for p in lm], dtype=np.float32)[:468]
    return rgb, coords_norm * np.array([w, h], dtype=np.float32)

def _visualize_face(rgb: np.ndarray, coords_px: np.ndarray,
                    local_scores: dict, global_score: float,
                    gt_score: float, filename: str):
    """
    IEEE-style figure:
      - Kiri : wajah + bounding box berwarna per organ (tanpa label teks)
      - Kanan: bar chart organ scores (serif font, clean spines)
    """
    organ_order = list(ORGAN_INDICES.keys())
    img_ann     = rgb.copy()
    h_img, w_img = rgb.shape[:2]
    thick = max(2, int(min(h_img, w_img) / 150))

    for organ in organ_order:
        pts = coords_px[ORGAN_INDICES[organ]]
        x1, y1 = pts[:, 0].min(), pts[:, 1].min()
        x2, y2 = pts[:, 0].max(), pts[:, 1].max()
        pw, ph = (x2 - x1) * 0.08, (y2 - y1) * 0.08
        x1 = max(0,     int(x1 - pw));  y1 = max(0,     int(y1 - ph))
        x2 = min(w_img, int(x2 + pw));  y2 = min(h_img, int(y2 + ph))
        cv2.rectangle(img_ann, (x1, y1), (x2, y2), _COLORS_BGR[organ], thick)

    fig, (ax_img, ax_bar) = plt.subplots(
        1, 2, figsize=(7.16, 3.2),
        gridspec_kw={'width_ratios': [1, 1.15]}
    )

    # ── Panel kiri ───────────────────────────────────────────────────────────
    ax_img.imshow(img_ann)
    ax_img.axis('off')
    ax_img.set_title('(a) Feature Bounding Boxes', loc='left', fontsize=9, pad=5)
    legend_patches = [
        mpatches.Patch(facecolor=_COLORS_HEX[o], linewidth=0,
                       label=_ORGAN_LABELS[o])
        for o in organ_order
    ]
    ax_img.legend(handles=legend_patches, loc='lower center',
                  ncol=3, fontsize=7.5, framealpha=0.92,
                  edgecolor='0.7', bbox_to_anchor=(0.5, -0.01))

    # ── Panel kanan ──────────────────────────────────────────────────────────
    bar_labels = [_ORGAN_LABELS[o] for o in organ_order]
    bar_scores = [local_scores.get(o, 0.0) for o in organ_order]
    bar_colors = [_COLORS_HEX[o] for o in organ_order]

    bars = ax_bar.bar(range(len(bar_labels)), bar_scores,
                      color=bar_colors, linewidth=0, width=0.55, zorder=3)

    for bar, sc in zip(bars, bar_scores):
        ax_bar.text(bar.get_x() + bar.get_width() / 2, sc + 0.05,
                    f'{sc:.2f}', ha='center', va='bottom', fontsize=8)

    ax_bar.axhline(global_score, color='black', ls='--', lw=0.9, zorder=4,
                   label=f'Global = {global_score:.2f}')
    ax_bar.axhline(gt_score,     color='0.45',  ls=':',  lw=0.9, zorder=4,
                   label=f'GT = {gt_score:.2f}')

    ax_bar.set_xlim(-0.5, len(bar_labels) - 0.5)
    ax_bar.set_ylim(1.0, 5.3)
    ax_bar.set_xticks(range(len(bar_labels)))
    ax_bar.set_xticklabels(bar_labels, rotation=20, ha='right', fontsize=8)
    ax_bar.set_ylabel('Beauty Score')
    ax_bar.set_title(f'(b) Individual Feature Prediction\n{filename}',
                     loc='left', fontsize=9, pad=5)
    ax_bar.legend(loc='upper left', handlelength=1.8)
    ax_bar.yaxis.grid(True, zorder=0)
    ax_bar.set_axisbelow(True)

    plt.tight_layout(pad=0.8)
    return fig

# ── Sample & jalankan ─────────────────────────────────────────────────────────
test_df_vis = pd.read_csv(TEST_CSV)
sample_rows = test_df_vis.sample(n=N_SAMPLES, random_state=RANDOM_SEED).reset_index(drop=True)
trained_model.eval()
saved_figs  = []

for _, row in sample_rows.iterrows():
    fname    = row['Filename']
    gt       = float(row['Rating'])
    img_path = os.path.join(IMAGE_DIR, fname)

    if not os.path.exists(img_path):
        print(f'\u26a0 File tidak ada: {img_path}')
        continue

    mp_result = _get_pixel_landmarks(img_path)
    if mp_result is None:
        print(f'\u26a0 Wajah tidak terdeteksi: {fname}')
        continue
    rgb_img, coords_px = mp_result

    if fname not in test_face_ds.filenames:
        print(f'\u26a0 {fname} tidak ada di test dataset')
        continue
    idx_ds = test_face_ds.filenames.index(fname)
    sample = test_face_ds[idx_ds]
    batch  = collate_faces([sample])

    with torch.no_grad():
        sg  = {k: v.to(device) for k, v in batch['subgraphs'].items()}
        out = trained_model(sg)

    local_scores = {o: out['local_scores'][o][0].item() for o in out['local_scores']}
    global_score = out['global_score'][0].item()

    print(f'\n{fname}  |  GT={gt:.2f}  Global={global_score:.2f}')
    for o, s in local_scores.items():
        print(f'  {o:<12}: {s:.3f}')

    fig = _visualize_face(rgb_img, coords_px, local_scores, global_score, gt, fname)
    save_path = f'{DRIVE_ROOT}/viz_{fname}'
    fig.savefig(save_path + '.pdf')
    fig.savefig(save_path + '.png')
    saved_figs.append(save_path + '.png')
    plt.show()
    plt.close(fig)

print(f'\n\u2713 {len(saved_figs)} visualisasi disimpan ke Drive (PDF + PNG).')
for p in saved_figs:
    print(f'  {p}')
````
